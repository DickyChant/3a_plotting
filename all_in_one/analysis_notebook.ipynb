## Interactive Histogram Exploration

Once you have histograms, you can explore them interactively.

```python
# Example: Create a simple histogram for demonstration
# This uses synthetic data to show the plotting functionality

import numpy as np

# Create a sample histogram
h_demo = ROOT.TH1F("h_demo", "Demo Histogram;Mass [GeV];Events", 50, 0, 500)
h_demo.FillRandom("gaus", 10000)

# Draw it
c = ROOT.TCanvas("c_demo", "", 600, 400)
h_demo.Draw("HIST")
c.Draw()

print(f"Demo histogram: {h_demo.GetEntries()} entries, mean = {h_demo.GetMean():.2f}")
```

## Using Distributed RDataFrame with Dask

For large datasets, you can use distributed RDataFrame with Dask.
This requires ROOT >= 6.26 with experimental distributed RDataFrame support.

```python
# Check if distributed RDataFrame is available
try:
    ROOT.RDF.Experimental.Distributed
    print("Distributed RDataFrame is available!")
    print("You can use utils.init_distributed_rdf() to get a distributed RDataFrame class.")
except AttributeError:
    print("Distributed RDataFrame is not available in this ROOT version.")
    print("Consider upgrading ROOT or using implicit multithreading instead.")
```
